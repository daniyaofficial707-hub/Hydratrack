```python
import pandas as pd
from datetime import datetime
import os
```

```python
file_name="water_log.csv"

if not os.path.exists(file_name):
     df=pd.DataFrame(columns=
      ["date","time","amount(ml)"])
     df.to_csv(file_name,index=False)
     print("file created!")
else:
  print("file already exists!")

```

```python
amount=int(input("enter water amount(ml):"))

now=datetime.now()
date=now.strftime("%y-%m-%d")
time=now.strftime("%h:%m:%s")

new_entry=pd.DataFrame([[date,time,amount]],columns=["date","time","amount(ml)"])
new_entry.to_csv(file_name, mode='a',header=False,index=False)

print("water intake recorded!")
```

```python
import pandas as pd
from datetime import datetime
df=pd.read_csv("water_log.csv")

#get today's data
today=datetime.now().strftime("%y-%m-%d")

#filter today's data
today_data=df[df["date"]==today]

#calculate total
total=today_data["amount(ml)"].sum()
print(f"total water intake today: {total}ml")
```

```python
print(df.columns)
```

```python
goal=2000 #2 liters
if total>=goal:
  print("goal achieved! you are hydrated likw a legend.")
else:
  remaining=goal-total
  print(f"you need {remaining} ml more to reach your goal")

```

```python
import matplotlib.pyplot as plt
#group by date
daily=df.groupby("date")["amount(ml)"].sum()

#plot
daily.plot(kind="bar")
plt.title("daily water intake")
plt.xlabel("date")
plt.ylabel("water(ml)")
plt.show()
```

```python
import matplotlib.pyplot as plt

labels=["Drank","Remaining"]
values=[total,max(0,goal-total)]

plt.bar(labels,values)
plt.title("today's hydration progress")
plt.ylabel("water(ml)")
plt.show()
```

```python
avg=df["amount(ml)"].mean()
print(f"average intake per entry:{avg:2f}ml")
```
